```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox
import papermill as pm
import subprocess
from datetime import datetime

date = datetime.now().strftime("%d-%m-%Y")

def run_faceAdd_file():
    name = entry_name.get()
    Email = entry_email.get()
    Department = entry_department.get()
    Entry_number = entry_number.get()
    
    if not name or not Email or not Department or not Entry_number:
             messagebox.showwarning("Input Error", "Please fill all details...")
             return
    
    confirm =messagebox.askyesno("Running", f"Starting Face collection process...")
    if confirm:  
        try:       
            pm.execute_notebook( "Face_Email_View_sample.ipynb", "Face_Email_View_sample.ipynb",
                                parameters={'name': name, 'Email': Email,'Department': Department, 'Entry_number': Entry_number})
      
            messagebox.showinfo("Success", "Face added successfully!")
            
        except Exception as e:
            messagebox.showerror("Error", f"Execution failed:\n{e}")
        

def run_recognition_file():        
        confirm =messagebox.askyesno("Running", "Starting Attendance process...")
        if confirm:    
            try:
                pm.execute_notebook("Face_Email_View_recognition.ipynb", "Face_Email_View_recognition.ipynb")
                
                messagebox.showinfo("Success", "Attendance Taken successfully!")
                
            except Exception as e:
                messagebox.showerror("Error", f"Execution failed:\n{e}")


def run_OpenCsv_file():
    csv_file = f"Attendance/Attendance_{date}.csv"
    try:
        subprocess.Popen(['start', csv_file], shell=True)
    except Exception as e:
        print(f"Error opening file: {e}")
        

def reset_entry(): 
    entry_name.delete(0, tk.END)                
    entry_email.delete(0, tk.END)
    entry_department.delete(0, tk.END)
    entry_department.set("Select department")
    entry_number.delete(0, tk.END)

departments = ["Computer Science Engineering","Mathematics &Com Engineering", "Electronics &Com Engineering","Electrical Engineering", "civil Engineering",
               "Mechanical Engineering","Management","Economic","Mathematics"]

def exit_app():
    confirm = messagebox.askyesno("Exit", "Are you sure you want to exit?")
    if confirm:
        root.destroy()
        
# GUI Setup
root = tk.Tk()
root.title("Student Attendance Management Dashboard")
root.geometry("450x550")
root.configure(bg="#d4f739")

# Main headline label
headline = tk.Label(root, text="Face Authentication based Attendance system", font=("Helvetica", 15, "bold"),
                    bg="#d4f739",fg="#800000")
headline.pack(pady=2)

tk.Label(root, text="Enter Your Name:", fg="#076834", font=("Helvetica", 15, "bold"), bg="#d4f739").pack(pady=5)
entry_name = tk.Entry(root, bg="#fdffb6", width=35)
entry_name.pack(pady=5)


tk.Label(root, text="Enter Your Email:", fg="#076834", font=("Helvetica", 15, "bold"), bg="#d4f739").pack(pady=5)
entry_email = tk.Entry(root, bg="#fdffb6", width=35)
entry_email.pack(pady=5)

# New row for department and Entry Numbers
row_frame = tk.Frame(root, bg="#d4f739")
row_frame.pack(pady=5)

tk.Label(row_frame, text="Department:", fg="#076834", font=("Helvetica", 15, "bold"), bg="#d4f739").grid(row=0, column=0, padx=10)
selected_dept = tk.StringVar()
entry_department = ttk.Combobox(row_frame,textvariable=selected_dept,values=departments,state="readonly",width=25)
entry_department.set("Select department")
entry_department.grid(row=1, column=0, padx=10)


tk.Label(row_frame, text="Entry Number:", fg="#076834", font=("Helvetica", 15, "bold"), bg="#d4f739").grid(row=0, column=1, padx=10)
entry_number = tk.Entry(row_frame, bg="#FFFFFF", width=27)
entry_number.grid(row=1, column=1, padx=10)


#Buttons
tk.Button(root, text="Reset", bg="#FFD700", width=8, height=1, font=("Helvetica", 9, "bold"), command=reset_entry).pack(pady=10)

tk.Button(root, text="Face Registration", bg="#1b960b", fg="white", width=35, height=1, font=("Helvetica", 10, "bold"), 
          command=run_faceAdd_file).pack(pady=10)

tk.Button(root, text="Take Attendance", bg="#2196F3", fg="white", width=22, height=2, font=("Helvetica", 10, "bold"),
          command=run_recognition_file).pack(pady=10)

tk.Button(root, text="View Attendance", bg="#736b73", fg="white", width=17, height=2, font=("Helvetica", 10, "bold"),
          command=run_OpenCsv_file).pack(pady=10)

tk.Button(root, text="Exit", bg="#f44336", fg="white", width=10, height=1, font=("Helvetica", 10, "bold"), 
          command=exit_app).pack(pady=10)

root.mainloop()
```
